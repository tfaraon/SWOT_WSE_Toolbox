```python
lon_point = 137.034
lat_point = -28.893
site_name = 'Belt Bay'

map_display = splot.plot_point_on_map(lon_point, lat_point)

#get z correction
mnt_z = stools.get_mnt_value(stools.load_mnt(path_mnt),lon_point, lat_point)
print(mnt_z)

map_display
```

### Dataset Quality

```python
qual_timeseries_394_100 = stools.extract_qual_timeseries_parallel(path, lon_point, lat_point, filter_pass='394', filter_resolution='100')
qual_timeseries_394_250 = stools.extract_qual_timeseries_parallel(path, lon_point, lat_point, filter_pass='394', filter_resolution='250')
qual_timeseries_435_100 = stools.extract_qual_timeseries_parallel(path, lon_point, lat_point, filter_pass='435', filter_resolution='100')
qual_timeseries_435_250 = stools.extract_qual_timeseries_parallel(path, lon_point, lat_point, filter_pass='435', filter_resolution='250')
qual_timeseries_52F = stools.extract_qual_timeseries_parallel(path, lon_point, lat_point, filter_tile='52F')
qual_timeseries_53F = stools.extract_qual_timeseries_parallel(path, lon_point, lat_point, filter_tile='53F')
qual_timeseries_102F = stools.extract_qual_timeseries_parallel(path, lon_point, lat_point, filter_tile='102F')
qual_timeseries_103F = stools.extract_qual_timeseries_parallel(path, lon_point, lat_point, filter_tile='103F')


data_dict = {       
    '394_100': qual_timeseries_394_100,
    '394_250': qual_timeseries_394_250,
    '435_100': qual_timeseries_435_100,
    '435_250': qual_timeseries_435_250}

# Plotting the timeseries
splot.plot_timeseries_multiple(data_dict, lon=lon_point, lat=lat_point, title=f'Timerseries at {site_name} ({lon_point};{lat_point})', var_name='qual')
print(stools.compute_qual_percentages(data_dict))

data_dict2 = {
    '52F': qual_timeseries_52F,
    '53F': qual_timeseries_53F,
    '102F': qual_timeseries_102F,
    '103F': qual_timeseries_103F}

# Plotting the timeseries
splot.plot_timeseries_multiple(data_dict2, lon=lon_point, lat=lat_point, title=f'Timerseries at {site_name} ({lon_point};{lat_point})', var_name='qual')
print(stools.compute_qual_percentages(data_dict2))
```
